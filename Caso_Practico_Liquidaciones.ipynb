```python
import secrets
clave_db = secrets.token_urlsafe(18)
import subprocess
sql = (f"CREATE USER IF NOT EXISTS 'alumno'@'localhost' IDENTIFIED BY '{clave_db}';"
       f"GRANT ALL PRIVILEGES ON santrix_db.* TO 'alumno'@'localhost';"
       "FLUSH PRIVILEGES;")
r = subprocess.run(["mysql", "-u", "root"], input=sql, capture_output=True, text=True)
print("Crear usuario con privilegios sobre santrix_db:", "OK" if r.returncode == 0 else r.stderr)
!mysql -u root -e "SHOW GRANTS FOR 'alumno'@'localhost';"
```

```python
import subprocess, os
os.environ["COMPOSER_ALLOW_SUPERUSER"] = "1"

def sh(comando, lineas=25):
    r = subprocess.run(comando, shell=True, capture_output=True, text=True,
                       cwd="/content/LIQUIDACION")
    salida = (r.stdout + r.stderr).strip().splitlines()
    print("\n".join(salida[-lineas:]))
    print("✅ OK" if r.returncode == 0 else f"❌ ERROR (código {r.returncode})")

sh("composer --version; node -v; npm -v")
```

```python
from pathlib import Path
p = Path("/content/LIQUIDACION")
env = p/".env"
if not env.exists():
    env.write_text((p/".env.example").read_text())
texto = env.read_text()
for k,v in {"DB_CONNECTION":"mysql",
 "DB_HOST":"127.0.0.1","DB_PORT":"3306",
 "DB_DATABASE":"santrix_db","DB_USERNAME":"alumno",
 "DB_PASSWORD":clave_db}.items():
    import re
    patron = rf"(?m)^{k}=.*$"
    texto = re.sub(patron, lambda m:f"{k}={v}", texto) if re.search(patron,texto) else texto+f"\n{k}={v}"
env.write_text(texto)
print(".env configurado")
```

```python
from pathlib import Path
prov = Path("/content/LIQUIDACION/app/Providers/AppServiceProvider.php")
codigo = prov.read_text()
if "forceScheme" not in codigo:
    codigo = codigo.replace(
        "public function boot(): void\n    {\n        //\n    }",
        "public function boot(): void\n    {\n"
        "        if (! in_array(request()->getHost(), ['127.0.0.1', 'localhost'])) {\n"
        "            \\Illuminate\\Support\\Facades\\URL::forceScheme('https');\n"
        "        }\n    }")
    prov.write_text(codigo)
print("HTTPS configurado" if "forceScheme" in prov.read_text() else "No se pudo modificar")
```

```python
from pathlib import Path
p = Path("/content/LIQUIDACION/app/Services/Anexo/AnexoExcelMapper.php")
c = p.read_text()
cambios = {
    "'FECHA DE ASIGNACION DE TRABAJO', 'FECHA DE ASIGNACION']": "'FECHA DE ASIGNACION DE TRABAJO', 'FECHA DE ASIGNACION', 'FECHA ASIGNACION']",
    "'ENTREGABLE DEL SERVICIO TIPO DE DOCUMENTO INFORME OTROS', 'TIPO DE ENTREGABLE']": "'ENTREGABLE DEL SERVICIO TIPO DE DOCUMENTO INFORME OTROS', 'TIPO DE ENTREGABLE', 'TIPO ENTREGABLE']",
    "'fecha_visita' => ['FECHA DE VISITA']": "'fecha_visita' => ['FECHA DE VISITA', 'FECHA VISITA']",
    "'FECHA DE DERIVACION DE ENTREGABLE SIGED CORRE', 'FECHA DE DERIVACION']": "'FECHA DE DERIVACION DE ENTREGABLE SIGED CORRE', 'FECHA DE DERIVACION', 'FECHA DERIVACION']",
}
n = 0
for a, b in cambios.items():
    if a in c:
        c = c.replace(a, b); n += 1
p.write_text(c)
print("Importador corregido:", n, "de 4 columnas")
sh("php -l app/Services/Anexo/AnexoExcelMapper.php")
```
